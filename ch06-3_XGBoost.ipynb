### 위스콘신 유방암 예측

```python
import pandas as pd

from sklearn import set_config
set_config(display="text")

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

import warnings
warnings.filterwarnings('ignore')
```

```python
dataset = load_breast_cancer()

x_features = dataset.data
y_label = dataset.target

cancer_df = pd.DataFrame(data= x_features, columns= dataset.feature_names)
cancer_df['target'] = y_label
cancer_df.head()
```

```python
print(dataset.target_names)
print(cancer_df['target'].value_counts())
```

```python
X_features = cancer_df.iloc[:,:-1]
y_label = cancer_df.iloc[:,-1]

# 전체 데이터 중 80%는 학습 데이터, 20%는 테스트용 데이터 추출
X_train, X_test, y_train, y_test = train_test_split(x_features
                                                  , y_label
                                                  , test_size=0.2, random_state=100)

# 위에서 만든 X_train, y_train을 다시 쪼개서 90%는 학습용으로, 10%는 검증용 데이터로 분리
X_tr, X_val, y_tr, y_val = train_test_split(X_train, y_train, test_size = 0.1, random_state = 100)

print(X_train.shape, X_test.shape)
print(X_tr.shape, X_val.shape)
```

### 파이썬 래퍼 XGBoost 적용

```python
import xgboost as xgb
from xgboost import plot_importance
import numpy as np

dtr   = xgb.DMatrix(data = X_tr,   label = y_tr) # 학습
dval  = xgb.DMatrix(data = X_val,  label = y_val) # 검증
dtest = xgb.DMatrix(data = X_test, label = y_test) # 테스트
```

```python
params = {'max_depth' : 3, 'eta' : 0.05, 'objective' : 'binary:logistic', 'eval_metric' : 'logloss'}

num_rounds = 400 #총 400번 반복 학습
```

```python
# 학습 데이터 셋은 'train' 또는 평가 데이터 셋은 'eval'로 명기
eval_list = [(dtr, 'train'), (dval, 'eval')]

# 하이퍼 파라미터와 early stopping 파라미터를 train() 함수의 파라미터 전달 -> 의사결정나무 생성
xgb_model = xgb.train(params                = params
                    , dtrain                = dtr
                    , num_boost_round       = num_rounds
                    , early_stopping_rounds = 50
                    , evals                 = eval_list)
```

```python
pred_probs = xgb_model.predict(dtest)
np.round(pred_probs[:10], 3)
```

```python
# 예측 확률이 0.5보다 크면 1, 그렇지 않으면 0으로 예측값을 결정하여 list 객체인 predicted에 저장
predicted = [1 if x > 0.5 else 0 for x in pred_probs]
predicted
```

### 성능 평가

```python
from sklearn.metrics import confusion_matrix, accuracy_score
from sklearn.metrics import precision_score, recall_score
from sklearn.metrics import f1_score, roc_auc_score

def get_clf_eval(y_test, pred = None, pred_probs = None) :
 confusion = confusion_matrix(y_test, pred)
 accuracy  = accuracy_score(y_test, pred)
 precision = precision_score(y_test, pred)
 recall    = recall_score(y_test, pred)
 f1        = f1_score(y_test, pred)

  # ROC_ACU 추가
 roc_acu   = roc_auc_score(y_test, pred_probs)
 print("오차 행렬")
 print(confusion)
 print(f'정확도 : {accuracy}\n'
       f'정밀도 : {precision}\n'
       f'재현율 : {recall}\n'
       f'f1-scroe : {f1}\n'
       f'ACU : {roc_acu}')
```

```python
get_clf_eval(y_test, predicted, pred_probs)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize = (10, 12))
plot_importance(xgb_model, ax=ax)
plt.savefig('xgb_feature_importance.tif', format ='tif', dpi=300)
```

### 사이킷런 래퍼 XGBoost의 적용

```python
from xgboost import XGBClassifier
```

```python
xgb_wrapper   = XGBClassifier(n_estimators  = 400
                            , learning_rate = 0.1
                            , max_depth     = 3).fit(X_train
                                                   , y_train)

w_preds       = xgb_wrapper.predict(X_test)
w_preds_proba = xgb_wrapper.predict_proba(X_test)[:,1]
```

```python
get_clf_eval(y_test, w_preds, w_preds_proba)
```

```python
# 조기종료(early stopping)을 제공

evals       = [(X_test, y_test)]

xgb_wrapper = XGBClassifier(n_estimators          = 400
                          , learning_rate         = 0.1
                          , max_depth             = 3
                          , eval_matric           = 'logloss'
                          , early_stopping_rounds = 50)

xgb_wrapper.fit(X_train
              , y_train
              , eval_set = evals
              , verbose  = True)

w_preds       = xgb_wrapper.predict(X_test)
w_preds_proba = xgb_wrapper.predict_proba(X_test)[:,1]
```

```python
get_clf_eval(y_test, w_preds, w_preds_proba)
```
