```python
# 훈련 데이터
x_train = train_df[["longitude", "latitude"]]
y_train = train_df[["label"]]
```

```python
# 테스트 데이터
x_test = test_df[["longitude", "latitude"]]
y_test = test_df[["label"]]
```

# 모델 생성

```python
from sklearn import tree
from sklearn import preprocessing
```

```python
# pyplot은 숫자로 표현된 레이블을 시각화할 수 있음.
# LabelEncoder로 레이블을 숫자로 변경.
le = preprocessing.LabelEncoder()
y_encoded = le.fit_transform(y_train)

y_encoded
```

```python
clf = tree.DecisionTreeClassifier(random_state=35)
clf.fit(x_train, y_encoded) # 학습실행 수자형만 가능
```

```python
# def display_decision_surface(clf=clf, x=x_train, y=y_encoded):
def display_decision_surface(clf, x, y):
    
    # 차트의 범위가 모든 학습 데이터를 포함하도록 설정
    x_min = x.longitude.min() - 0.01 # 시각화 여유공간 잘리지 않도록
    x_max = x.longitude.max() + 0.01 # 시각화 여유공간 잘리지 않도록

    y_min = x.latitude.min() - 0.01 # 시각화 여유공간 잘리지 않도록
    y_max = x.latitude.max() + 0.01 # 시각화 여유공간 잘리지 않도록


    # 파라미터 설정
    n_classes = len(le.classes_)
    plot_colors = "rywb"
    plot_step = 0.001
   
    # 학습 데이터를 차트에 표시
    xx, yy = np.meshgrid(np.arange(x_min, x_max, plot_step), 
             np.arange(y_min, y_max, plot_step))

    z = clf.predict(np.c_[xx.ravel(), yy.ravel()])
    z = z.reshape(xx.shape)
    cs = plt.contourf(xx, yy, z, cmap=plt.cm.RdYlBu)
        

    for i, color in zip(range(n_classes), plot_colors):
        idx = np.where(y==i)
        plt.scatter(  x.loc[idx].longitude
                    , x.loc[idx].latitude
                    , c = color
                    , label = le.classes_[i]
                    , cmap=plt.cm.RdYlBu
                    , edgecolors='black', s=200)

    # 차트 제목
    plt.title('Decision surface of a desicsion tree', fontsize = 16)

    # 범례
    plt.legend(bbox_to_anchor=(1.05, 0.25), loc='lower left', fontsize=14)

    # x축의 이름과 폰트 크기 설정
    plt.xlabel('longitude', fontsize = 16) # 경도

    # y축의 이름과 폰트 크기 설정
    plt.ylabel('latitude', fontsize = 16) # 위도
     
    # 차트 크기 설정
    plt.rcParams['figure.figsize'] = [7 , 5] 
        
    # x축 좌표상의 폰트 크기 설정
    plt.rcParams['xtick.labelsize'] = 14 # x축 좌표
    
    # y축 좌표상의 폰트 크기 설정
    plt.rcParams['ytick.labelsize'] = 14 # y축 좌표
     
    # 차트 그리기
    plt.show()    
```

```python
# 함수 호출
display_decision_surface(clf, x_train, y_encoded)
```

## 파라미터 설정한 모델의 결정 표면 시각화

```python
clf = tree.DecisionTreeClassifier( max_depth         = 4   # 트리의 최대 한도 깊이 4개
                                 , min_samples_split = 2   # 자식 노드를 갖기 위한 최소한의 데이터 갯수
                                 , min_samples_leaf  = 2   # 맨 마지막 끝 노드의 최소 데이터 갯수
                                 , random_state      = 70) # 여러번 실행해도, 파라미터가 같은 경우, 결과가 항상 같게 만들어주는 파라미터

clf.fit(x_train, y_encoded.ravel())

display_decision_surface(clf, x_train, y_encoded)
```
